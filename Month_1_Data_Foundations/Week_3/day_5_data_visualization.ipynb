```python
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David",
             "Emma", "Frank", "Grace", "Henry",
             "Ivy", "Jack"],

    "Department": ["CS", "IT", "CS", "IT", "CS",
                   "IT", "CS", "IT", "CS", "IT"],

    "Gender": ["F", "M", "M", "M", "F",
               "M", "F", "M", "F", "M"],

    "Study_Hours": [5, 6, 3, 4, 7, 2, 8, 4, 6, 3],

    "Score": [85, 90, 65, 75, 95, 55, 98, 72, 88, 68],

    "Absences": [2, 1, 6, 4, 0, 8, 1, 5, 2, 7]
})
df
```

```python
#Display the average Score by Department.
avg_score=df.groupby("Department")["Score"].mean().reset_index()
avg_score
```

```python
plt.bar(avg_score.Department,avg_score.Score)
```

```python
#Display the average Score by Gender.
avg_score_gender=df.groupby("Gender")["Score"].mean().reset_index()
plt.bar(avg_score_gender.Gender,avg_score_gender.Score)
```

```python
#Display the distribution of Score using five bins and black edges.
plt.hist(
    df["Score"],
    bins=5,
    edgecolor="black"
)
plt.ylabel("studeent_number")
plt.xlabel("Scores")
```

```python
#Create two histograms using 3 and 10 bins. Explain their differences.
plt.hist(
    df["Score"],
    bins=3,
    edgecolor="black"
)
```

```python
plt.hist(
    df["Score"],
    bins=10,
    edgecolor="black"
)
```

```python
#Visualize Study_Hours against Score.
plt.scatter(
    df["Study_Hours"],
    df["Score"],
    color="purple",
    s=50,
    alpha=0.8

)

plt.xlabel("Study_hours")
plt.ylabel("Score")
```

```python
#Visualize Absences against Score and explain the relationship.
plt.scatter(
    df["Absences"],
    df["Score"],
    color="black",
    s=60,
    alpha=0.8

)
plt.xlabel("Absences")
plt.ylabel("Score")

#the more abasences the student, the student score decrease. It is negative correlection
```

```python
#Display average Study_Hours by Department.
import plotly.express as px

avg_sh_dep=df.groupby("Department")["Study_Hours"].mean().reset_index()
avg_sh_dep
```

```python
import plotly.express as px
fig=px.bar(
    df,
    x=avg_sh_dep["Department"],
    y=avg_sh_dep["Study_Hours"]
)
fig.show()
```

```python
#Display the distribution of Study_Hours.
fig=px.histogram(
    df,
    x="Study_Hours",
    nbins=5,
    title="Distribution of study hours"

)
fig.update_traces(
    marker_line_color="black",
    marker_line_width=1
)
fig.show()
```

```python
#Plot Study_Hours against Score, color by Gender, display Name and Absences on hover.
fig=px.scatter(
    df,
    x="Study_Hours",
    y="Score",
    color="Gender",
    hover_name="Name",
    hover_data=["Absences"]


)
fig.show()
```

```python
import pandas as pd
hotels = pd.DataFrame({
    "Booking_ID": range(101, 113),

    "Date": pd.date_range("2026-11-01", periods=12),

    "City": [
        "Bangkok", "Phuket", "Chiang Mai",
        "Bangkok", "Phuket", "Chiang Mai",
        "Bangkok", "Phuket", "Chiang Mai",
        "Bangkok", "Phuket", "Chiang Mai"
    ],

    "Price_Per_Night": [
        1500, 2500, 900, 1800, 3200, 1200,
        2200, 2800, 1000, 1600, 3500, 1400
    ],

    "Nights": [
        2, 4, 3, 1, 5, 2,
        3, 4, 2, 2, 6, 3
    ],

    "Rating": [
        4.2, 4.7, 3.8, 4.3, 4.9, 4.0,
        4.5, 4.6, 3.9, 4.1, 4.8, 4.4
    ],

    "Cancelled": [
        False, False, True, False,
        False, False, True, False,
        False, False, True, False
    ]
})

hotels
```

```python
hotels.shape
```

```python
hotels.dtypes
```

```python
hotels.duplicated()
```

```python
hotels.isna().sum()
```

```python
condition=hotels[hotels["Price_Per_Night"]<0]
condition
```

```python
check_night=hotels[hotels["Nights"]<0]
check_night
```

```python
check_rating=(hotels["Rating"]>=0)| (hotels["Rating"]<=5)
check_rating
```

```python
hotels["Booking_Value"]=hotels["Price_Per_Night"]*hotels["Nights"]
hotels["Booking_Value"]
```

```python
hotels
```

```python
hotels["Revenue"]=hotels["Booking_Value"]
hotels
```

```python
hotels.loc[hotels["Cancelled"]==1,"Revenue"]=0
```

```python
hotels
```

```python
#Create a line chart showing daily revenue.
daily_revenue=hotels.groupby("Date")["Revenue"].sum().reset_index()
daily_revenue
```

```python
import matplotlib.pyplot as plt

plt.plot(
    daily_revenue["Date"],
    daily_revenue["Revenue"]
)
```

```python
total_revenue_city=hotels.groupby("City")["Revenue"].sum()
total_revenue_city
```

```python
bar_chart=plt.bar(
    total_revenue_city.index,total_revenue_city.values
)
```

```python
avg_price_city=hotels.groupby("City")["Price_Per_Night"].mean()
avg_price_city
```

```python
bar_chart_city=plt.bar(
    avg_price_city.index,avg_price_city.values
)
```

```python
#Create a histogram showing the distribution of hotel prices.
histogram=plt.hist(
    hotels["Price_Per_Night"],
    bins=5,
    edgecolor="black"
)
plt.title("Distribution of Hotel Prices")
plt.xlabel("Price Per Night")
plt.ylabel("Frequency")
plt.show()
```

```python
#Create a scatter plot showing Price_Per_Night against Rating.
colors = hotels["City"].map({
    "Bangkok": "blue",
    "Phuket": "red",
    "Chiang Mai": "green"
})

plt.scatter(
    hotels["Price_Per_Night"],
    hotels["Rating"],
    color=colors,
    s=80,
    alpha=0.6
)
```

```python
hotels
```

```python
cancellation_percentages=hotels.groupby("City")["Cancelled"].mean()*100
cancellation_percentages
```

```python
plt.bar(
  cancellation_percentages.index,
  cancellation_percentages.values
)

plt.title("Cancellation Percentage by City")
plt.xlabel("City")
plt.ylabel("Cancellation Percentage (%)")
plt.show()
```
