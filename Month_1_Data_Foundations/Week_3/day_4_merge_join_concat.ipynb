```python
import pandas as pd
import numpy as np

students = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104],
    "Name": ["Alice", "Bob", "Charlie", "David"],
    "Age": [20, 21, 22, 23]
})

scores = pd.DataFrame({
    "Student_ID": [101, 102, 103, 105],
    "Score": [85, 90, 75, 95]
})
```

```python
result=pd.merge(
    students,
    scores,
    on="Student_ID"
)
result
```

```python
pd.merge(
    students,
    scores,
    on="Student_ID",
    how="left"
)
```

```python
students2 = pd.DataFrame({
    "Student_ID": [101, 102, 103],
    "Name": ["Alice", "Bob", "Charlie"]
})

results2 = pd.DataFrame({
    "ID": [101, 102, 103],
    "Score": [85, 90, 75]
})
```

```python
result=pd.merge(
    students2,
    results2,
    left_on="Student_ID",
    right_on ="ID"
)
result
```

```python
import pandas as pd
profiles = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4, 5],
    "Name": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "Department": ["CS", "IT", "CS", "IT", "CS"]
})

exam_results = pd.DataFrame({
    "Student_ID": [1, 2, 3, 5, 6],
    "Final_Score": [85, 90, 78, 95, 88]
})

study = pd.DataFrame({
    "Student_ID": [1, 2, 3, 4, 5],
    "Study_Hours": [5, 6, 3, 4, 7]
})
```

```python
all_students=profiles.set_index("Student_ID").join(
    [
    exam_results.set_index("Student_ID"),
     study.set_index("Student_ID")
     ],
    how="outer"
)
all_students
```

```python
students=pd.merge(
    profiles,
    exam_results,
    on="Student_ID",
    how="outer"
)
students=pd.merge(
    students,
    study,
    on="Student_ID",
    how="outer"
)
students
```

```python
#Identify students who have missing Final_Score values.
missing_final_students=students[students["Final_Score"].isna()]
missing_final_students
```

```python
#Calculate the average Final_Score by Department.
dep_avg_final=students.groupby("Department")["Final_Score"].mean()
dep_avg_final
```

```python
#Find the student with the highest Final_Score and display their name
students.sort_values("Final_Score",ascending=False).head(1)
```
