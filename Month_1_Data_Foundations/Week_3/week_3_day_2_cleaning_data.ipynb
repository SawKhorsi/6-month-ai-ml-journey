```python
import numpy as np
import pandas as pd
data={
    "Name":[
        "Alice", "Bob", "Charlie", "Alice",
        "David", "Emma", "Frank", "Grace"
    ],
    "Age":[
        "20", "21", "unknown", "20",
        "-5", "22", "23", "24"
    ],
    "Score":[
         "85", "90", "75", "85",
        "150", "88", "N/A", "95"
    ],
    "City":[
        "Bangkok", " bangkok ", "Chiang Mai",
        "Bangkok", "BANGKOK", "Chiang mai",
        "Phuket", "phuket"
    ]
}

df=pd.DataFrame(data)
df
```

```python
clean_df=df.copy()
```

```python
df["Name"]=df["Name"].str.strip().str.capitalize()
```

```python
df["City"]=df["City"].str.strip().str.capitalize()
```

```python
df
```

```python
clean_df=df.drop_duplicates(subset=["Name"])
clean_df
```

```python
clean_df.loc[
    (clean_df["Score"] < 0) |
    (clean_df["Score"] > 100),
    "Score"
] = np.nan

```

```python
clean_df=clean_df.dropna(subset=["Score"])
clean_df
```
