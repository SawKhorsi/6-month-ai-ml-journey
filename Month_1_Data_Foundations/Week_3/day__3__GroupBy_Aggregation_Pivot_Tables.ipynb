```python
import pandas as pd

data={
    "Name":["Alice", "Bob", "Charlie", "David",
             "Emma", "Frank", "Grace", "Henry"],
    "Gender": ["F", "M", "M", "M", "F", "M", "F", "M"],
    "City": ["Bangkok", "Bangkok", "Phuket", "Bangkok",
             "Phuket", "Phuket", "Bangkok", "Bangkok"],
    "Department": ["CS", "IT", "CS", "CS",
                   "IT", "IT", "CS", "IT"],
    "Score": [85, 90, 75, 80, 95, 70, 88, 82],
    "Study_Hours": [5, 6, 3, 4, 7, 2, 5, 4]

}

df=pd.DataFrame(data)
df
```

```python
df.groupby("City")["Score"].mean()
```

```python
df.groupby("City")["Score"].count()
```

```python
df.groupby("City")["Score"].size()
```

```python
df.groupby(["City","Gender"])["Score"].mean()
```

```python
result=df.groupby(["City","Gender"])["Score"].mean().reset_index()
result
```

```python
df.groupby(["City","Gender"])["Score"].agg(["mean","min","max"])
```

```python
df.columns
```

```python
df.groupby("Department").agg({
    "Score":["mean","min","max"],
    "Study_Hours":"mean"
})
```

```python
result=df.groupby("Department").agg(
    Average_Score=("Score","mean"),
    Minimum_Score=("Score","min"),
    Maximum_Score=("Score","max"),
    Average_StudyHours=("Study_Hours","mean")
)
result
```

```python
result.sort_values("Average_Score",ascending=False)

```

```python
pd.pivot_table(
    df,
    index="City",
    columns="Gender",
    values="Score",
    aggfunc="mean"
)
```

```python
df.groupby(["City","Gender"])["Score"].mean()
```

```python
df.groupby(["City","Gender"])["Score"].mean().unstack()
```

```python
pd.pivot_table(
    df,
    index="City",
    columns="Gender",
    values="Score",
    aggfunc="mean",
    margins=True
)
```
