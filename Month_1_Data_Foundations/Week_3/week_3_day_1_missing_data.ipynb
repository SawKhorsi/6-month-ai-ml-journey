```python
import pandas as pd
import numpy as np
data={
    "Name":["Alice","Bob","Charlie","David","Eliz"],
    "Age": [20, 21, np.nan, 22, 20],
    "Study_Hours": [5, np.nan, 3, 6, 4],
    "Score": [85, 90, 75, np.nan, 88],
    "City": ["Bangkok", "Bangkok", np.nan, "Chiang Mai", "Bangkok"]

}
df=pd.DataFrame(data)
df
```

```python
df[df.isna().any(axis=1)]
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().sum().sum()
```

```python
df[df.isna().any(axis=1)>=1]
```

```python
clean_df=df.copy()
clean_df
```

```python
clean_df["Study_Hours"]=clean_df["Study_Hours"].fillna(clean_df["Study_Hours"].mean ())
clean_df["Age"]=clean_df["Age"].fillna(clean_df["Age"].median())
clean_df["City"]=clean_df["City"].fillna(clean_df["City"].mode()[0])
clean_df=clean_df.dropna(subset=["Score"])

```

```python
clean_df
```

```python
print("Shape:", clean_df.shape)
print("Missing values:", clean_df.isna().sum().sum())
```
